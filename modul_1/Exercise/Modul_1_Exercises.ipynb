## Exercise 2 – Investigating Int and Float Memory Size

```python
# Nama: Rico Shandika Jovial Agista
# NIM: 202410370110433

import sys

# Integer
data1 = 123456789
data2 = 1234567890
print('Value =', data1, '| Type =', type(data1), '| Size =', sys.getsizeof(data1), 'bytes')
print('Value =', data2, '| Type =', type(data2), '| Size =', sys.getsizeof(data2), 'bytes')

# Float
f1 = 12345.6789
f2 = 1234567890.123456
print('Value =', f1, '| Type =', type(f1), '| Size =', sys.getsizeof(f1), 'bytes')
print('Value =', f2, '| Type =', type(f2), '| Size =', sys.getsizeof(f2), 'bytes')
```

## Exercise 8 – Analyzing a Recursive Function

```python
# Nama: Rico Shandika Jovial Agista
# NIM: 202410370110433

def total_sum(n):
    print(f"Current n: {n}")
    if n == 1:
        return 1
    else:
        return n + total_sum(n - 1)

result = total_sum(3)
print("Final Result:", result)
```
