## 2. Menghitung final score

Final score dihitung dengan rumus:

`final_score = (assignment * 0.4) + (exam * 0.6)`

Fungsi berikut membuat data baru dengan tambahan nilai final score.

```python
def calculate_final_scores(data):
    hasil = []
    for student in data:
        nama = student[0]
        assignment = student[1]
        exam = student[2]
        final_score = (assignment * 0.4) + (exam * 0.6)
        hasil.append([nama, assignment, exam, final_score])
    return hasil


all_updated = calculate_final_scores(raw_grades)

print("Data setelah final score dihitung:")
for student in all_updated:
    print(student)
```

## 5. Menghitung jumlah siswa dengan rekursi

Fungsi ini menghitung jumlah elemen tanpa menggunakan loop dan `len()`. Fungsi memanggil dirinya sendiri sampai indeks berada di luar data.

```python
def count_students_recursive(data, index=0):
    try:
        _ = data[index]
    except IndexError:
        return 0
    return 1 + count_students_recursive(data, index + 1)


total_students = count_students_recursive(selected_tuple)
print(f"Jumlah {list_label} students:", total_students)
```

## 6. Verifikasi nama siswa dengan try-except-else

Jika nama tidak ditemukan, fungsi memunculkan `ValueError` lalu menampilkan pesan error. Jika nama ditemukan, bagian `else` dijalankan.

```python
def verify_student(name, data, label):
    try:
        if name not in data:
            raise ValueError(
                f"Error: Student name '{name}' not found in the {label.lower()} list!"
            )
    except ValueError as error:
        print(error)
    else:
        print(f"Student Found '{name}'")


print("Scenario A - nama yang ada di tuple")
if selected_tuple:
    verify_student(selected_tuple[0], selected_tuple, list_label)
else:
    print("Tuple kosong")

print("\nScenario B - nama yang tidak ada di tuple")
if list_label == "Passed":
    verify_student("Nisrina", selected_tuple, list_label)
else:
    verify_student("Amel", selected_tuple, list_label)
```
